```python
# Install necessary packages
!pip install transformers --quiet
!pip install torch --quiet
!pip install gradio --quiet

from transformers import pipeline
import gradio as gr

# Load IBM Granite instruction-tuned model
pipe = pipeline(
    "text-generation",
    model="ibm-granite/granite-3.3-2b-instruct"
)

# Initialize conversation memory
conversation_history = []

# Function to generate chatbot response with formatted sections
def chatbot(user_input, user_type):
    global conversation_history

    # Add user message to history
    conversation_history.append(f"User ({user_type}): {user_input}")

    # Build prompt including conversation history and project description
    project_description = """
The Personal Finance Chatbot project aims to develop an intelligent conversational AI system
that leverages IBM's generative AI models and Watson services to provide personalized financial guidance.
It will answer user questions about savings, taxes, and investments, offer budget summaries,
and suggest actionable spending insights. It adapts tone and complexity for different demographics
(Student vs. Professional)."""

    expected_solutions = """
Expected Solutions:
1) Personalized Financial Guidance
2) AI-Generated Budget Summaries
3) Spending Insights and Suggestions
4) Demographic-Aware Communication
5) Conversational NLP Experience"""

    prompt = project_description + "\n" + expected_solutions + "\nConversation History:\n"
    prompt += "\n".join(conversation_history)
    prompt += f"\nAssistant: Respond with clear sections: Advice, Budget Summary, Spending Insights, tailored for a {user_type}."

    # Generate response
    output = pipe(prompt, max_new_tokens=350, do_sample=True, temperature=0.7)
    answer = output[0]["generated_text"]

    # Extract assistant response
    if "Assistant:" in answer:
        answer = answer.split("Assistant:")[-1].strip()

    # Add assistant response to history
    conversation_history.append(f"Assistant: {answer}")

    return answer

# Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("## Personal Finance Chatbot (IBM Granite) with Formatted Responses")

    with gr.Row():
        user_type = gr.Radio(["Student", "Professional"], label="Select your user type")

    user_input = gr.Textbox(label="Enter your question or prompt", lines=3)
    output_box = gr.Textbox(label="Chatbot Response", lines=15)
    submit_btn = gr.Button("Generate Response")
    reset_btn = gr.Button("Reset Chat")

    submit_btn.click(fn=chatbot, inputs=[user_input, user_type], outputs=[output_box])

    def reset_chat():
        global conversation_history
        conversation_history = []
        return ""

    reset_btn.click(fn=reset_chat, inputs=[], outputs=[output_box])

# Launch Gradio interface
demo.launch()
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("ibm-granite/granite-3.3-2b-instruct")
model = AutoModelForCausalLM.from_pretrained("ibm-granite/granite-3.3-2b-instruct")
messages = [
    {"role": "user", "content": "Who are you?"},
]
inputs = tokenizer.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=40)
print(tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```
